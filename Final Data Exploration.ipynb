## Data Generation and Validation Summary

Three Python scripts were used to prepare and validate the Medici Bank transaction dataset.

### 1. Initial historical data generation

The `generate_historical_data.py` script created 20,000 synthetic, historically themed banking transactions covering 1390–1440. The dataset includes deposits, withdrawals, loans, loan repayments, bills of exchange, operating expenses, alum trading, war financing, and the Council of Constance ransom payment.

### 2. Dataset expansion and embezzlement trail

The `generate_additional_data.py` script added 60,000 legitimate transactions and 230 intentionally suspicious operating-expense transactions. These suspicious payments:

- Total 95,610 florins
- Occur between January 3, 1420 and December 26, 1424
- Originate exclusively from the Florence branch
- Are paid to the fictitious vendor `Ser Benedetto Forniture`
- Use repetitive, mostly round payment amounts and occur at unusually regular intervals

After expansion, the dataset contains 80,230 transactions.

### 3. Data validation

The `validate_transactions.py` script confirmed that both the CSV and JSON datasets are structurally valid. All 80,230 transactions passed the accounting validation with no errors.

- Total debits: 21,612,302,640.94 florins
- Total credits: 21,612,302,640.94 florins
- Debit-credit difference: 0.00 florins
- CSV validation: Passed
- JSON validation: Passed
- Date coverage: 1390–1440, spanning 51 years

The equality of total debits and credits confirms that the dataset follows double-entry accounting principles.

## 7. CSV — Balancing Rule Check

- **Purpose:** Verify every transaction satisfies double-entry accounting before ingestion applies the same rule.
- **Why:** The data contract (Cleaned Transaction, `data-contracts.md`) requires `debit_amount == credit_amount + credit_amount_2` as the validation rule enforced before a row becomes a Cleaned Transaction; rows that fail are logged and excluded. Confirming this in raw data first tells the team what exclusion rate to expect.
- **Method:** Computed `abs(debit_amount - (credit_amount + credit_amount_2.fillna(0))) < 0.01` across all rows (tolerance used instead of exact equality due to float rounding), then counted True/False.
- **Findings:** All 80,230 rows balance; `True: 80230`, zero failures. The contract's exclusion logic won't drop any rows on this dataset; every transaction is already double-entry-clean.

```python
balanced = (df_csv["debit_amount"] - (df_csv["credit_amount"] + df_csv["credit_amount_2"].fillna(0))).abs() < 0.01
balanced.value_counts()
```

## 11. CSV — credit_account_2 / credit_amount_2  dealing with Nulls

- **Purpose:** Confirm whether `credit_account_2` and `credit_amount_2` are null together consistently.
- **Why:** The data contract defines an asymmetric rule for these two fields: blank `credit_amount_2` → `0` (never `None`), while blank `credit_account_2` → `None`. Before ingestion applies that transformation, need to confirm the raw data's null pattern is consistent to start from.
- **Method:** `(credit_account_2.isnull() == credit_amount_2.isnull()).value_counts()`.
- **Findings:** `True: 80230` in the raw CSV, the two columns are null together on every row, with no row having one filled and the other blank. Clean starting point: ingestion's asymmetric transform (amount → `0`, account → stays `None`) can be applied without needing to handle any misaligned edge cases first. Also confirmed via `.unique()`: `credit_account_2` only takes 3 values: `NaN`, `Interest Income`, `Exchange Fee Revenue` 

```python
# Are the two null columns the same; does account_2 being null match amount_2 being null?
(df_csv["credit_account_2"].isnull() == df_csv["credit_amount_2"].isnull()).value_counts()
```

## 12. CSV — Transaction Counts by Year

- **Purpose:** Count the number of transactions recorded during each year in the dataset.
- **Why:** Transaction counts by year are a required Phase 1 summary and help confirm that the dataset covers the complete historical period without unexpected missing years.
- **Method:** Derived a `year` column from the converted `date` column using `.dt.year`, grouped the records by year, and counted the number of transactions in each group.
- **Findings:** The dataset contains transactions for all 51 years from 1390 through 1440. The yearly counts sum to the complete dataset total of 80,230 transactions. The highest annual count occurs in 1424 with 1,651 transactions, while the lowest occurs in 1433 with 1,479 transactions. The relatively consistent annual counts show that the data is distributed across the full historical period rather than concentrated in only a few years.

```python
df_csv["year"] = df_csv["date"].dt.year
transactions_by_year = df_csv.groupby("year").size()
print(transactions_by_year.to_string())
```

## 13. CSV — Total Florin Volume by Branch and Transaction Type

- **Purpose:** Calculate the total florin value of transactions for every branch and transaction-type combination.
- **Why:** This summary is a required Phase 1 deliverable and shows which activities contribute the greatest financial volume within each branch. It also provides a baseline for the branch-level KPI calculations developed in later phases.
- **Method:** Grouped the transactions by `branch` and `type`, summed `debit_amount` for each group, and reshaped the result into a table with branches as rows and transaction types as columns. `debit_amount` was used because it represents the complete value of each balanced transaction; adding debit and credit amounts would count the same transaction twice.
- **Findings:** Florence has the largest total transaction volume, driven primarily by approximately 12.11 billion florins in war-financing activity. Rome has the second-largest volume, led by approximately 5.44 billion florins in deposits. Constance has only one transaction—the 35,000-florin ransom payment; so its total volume is substantially smaller than the operating branches. The results demonstrate that transaction count and transaction value measure different aspects of branch activity: Rome has the most transactions, while Florence has the greatest total florin volume.

```python
transactions_by_branch = df_csv.groupby("branch").size()
print(transactions_by_branch)
```

```python
volume_by_branch_and_type = (
    df_csv.groupby(["branch", "type"])["debit_amount"]
    .sum()
)
print(volume_by_branch_and_type.to_string())
```

```python
volume_table = (
    df_csv.groupby(["branch", "type"])["debit_amount"]
    .sum()
    .unstack(fill_value=0)
)
volume_table.style.format("{:,.2f}")
```
